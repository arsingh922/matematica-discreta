```python
import pandas as pd

df = pd.read_csv('WHO.csv')
print(list(df.columns))

# Let's inspect the first few rows of the country and the 4th column
fert_col = df.columns[3]
print(f"Fertility column: {fert_col}")

top_10 = df[['Country', fert_col]].sort_values(by=fert_col, ascending=False).head(10)
print("\nTop 10 countries:")
print(top_10)

mean_fert = df[fert_col].mean()
above_mean = (df[fert_col] > mean_fert).sum()
below_mean = (df[fert_col] < mean_fert).sum()

print(f"\nMean: {mean_fert}")
print(f"Above mean: {above_mean}")
print(f"Below mean: {below_mean}")
```
